## 1. Introducción a las Redes Neuronales Artificiales

Las Redes Neuronales Artificiales (ANN) son un subcampo del aprendizaje profundo (deep learning) inspirado en las neuronas biológicas. Los conceptos principales incluyen:
*   **Perceptrón:** La forma más simple de una red neuronal, que consta de nodos de entrada, pesos, una suma ponderada y una función de activación.
*   **Perceptrón Multicapa (MLP):** Redes con múltiples capas (entrada, oculta y salida) que pueden resolver problemas complejos no lineales.

## 2. Predicción de Tipos de Vino con Keras

El proyecto principal consiste en utilizar un conjunto de datos de vinos tintos y blancos para predecir el tipo de vino basándose en sus propiedades químicas.

### Carga de los Datos
El conjunto de datos se carga utilizando Pandas desde el Repositorio de Machine Learning de la UCI.

```python
import pandas as pd

# Leer datos del vino blanco
white = pd.read_csv("winequality-white.csv", sep=';')

# Leer datos del vino tinto
red = pd.read_csv("winequality-red.csv", sep=';')
```

### Exploración de los Datos
Podemos inspeccionar los datos utilizando `.info()`, `.describe()` y visualizando las distribuciones (por ejemplo, el contenido de alcohol) utilizando Matplotlib y Seaborn.

```python
# Información sobre el vino blanco
print(white.info())

# Información sobre el vino tinto
print(red.info())
```

## 3. Preprocesamiento de Datos
Para preparar los datos para la red neuronal, debemos:
1.  **Etiquetar los datos:** Asignar `1` al vino tinto y `0` al vino blanco.
2.  **Concatenar:** Combinar ambos conjuntos de datos en uno solo.
3.  **Dividir:** Utilizar `train_test_split` de Scikit-Learn.
4.  **Estandarizar:** Escalar las características a un rango común.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Añadir columna 'type' (tipo)
red['type'] = 1
white['type'] = 0

# Añadir los vinos blancos a los tintos
wines = pd.concat([red, white], ignore_index=True)

# Definir las características (X) y el objetivo (y)
X = wines.iloc[:, 0:11]
y = wines.type

# Dividir los datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# Estandarizar los datos
scaler = StandardScaler().fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

## 4. Modelado de los Datos
Se crea un modelo Secuencial con una capa de entrada, una capa oculta y una capa de salida.

```python
from keras.models import Sequential
from keras.layers import Dense

# Inicializar el constructor del modelo
model = Sequential()

# Añadir la capa de entrada
model.add(Dense(12, activation='relu', input_shape=(11,)))

# Añadir una capa oculta
model.add(Dense(8, activation='relu'))

# Añadir la capa de salida
model.add(Dense(1, activation='sigmoid'))
```

## 5. Compilación y Entrenamiento del Modelo
El modelo se compila con un optimizador (`adam`) y una función de pérdida (`binary_crossentropy`).

```python
model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
                   
# Entrenar el modelo
model.fit(X_train, y_train, epochs=20, batch_size=32, verbose=1)
```

## 6. Evaluación del Modelo
Después del entrenamiento, se evalúa el rendimiento del modelo en el conjunto de prueba.

```python
# Predecir los valores
y_pred = model.predict(X_test)

# Evaluar el modelo
score = model.evaluate(X_test, y_test, verbose=1)
print("Loss & Accuracy:", score)
```

## 7. Predicción de la Calidad del Vino
El proyecto se puede ampliar para predecir la *calidad* del vino (un problema de regresión o clasificación multiclase). Para esto se puede:
*   **Preprocesamiento:** Usar la columna `quality` como objetivo.
*   **Ajuste del Modelo (Fine-Tuning):** Añadir más capas o unidades ocultas para mejorar la precisión.

Este tutorial proporciona una base sólida para crear modelos de aprendizaje profundo en Python usando Keras.

### 7.1 Preparación de Datos para Predecir la Calidad

Ahora vamos a utilizar la columna `quality` como nuestro objetivo (target). Al predecir un valor numérico (puntuación del vino del 0 al 10), este se convierte en un problema de **Regresión**.

```python
# El objetivo ahora es la columna 'quality'
y_quality = wines.quality

# Dividir nuevamente los datos usando el nuevo objetivo (X no cambia, siguen siendo las 11 características)
X_train_q, X_test_q, y_train_q, y_test_q = train_test_split(X, y_quality, test_size=0.33, random_state=42)

# Estandarizar los datos usando un nuevo scaler
scaler_q = StandardScaler().fit(X_train_q)
X_train_q = scaler_q.transform(X_train_q)
X_test_q = scaler_q.transform(X_test_q)
```

### 7.2 Modelo para Regresión de Calidad

Como es un problema de regresión, la capa de salida ya no tendrá la activación `sigmoid`, sino que tendrá 1 nodo y activación lineal (por defecto) para poder predecir cualquier valor numérico continuo. La función de pérdida (loss) típica para regresión es el Error Cuadrático Medio (`mse`).

```python
model_q = Sequential()

# Capa de entrada y primera capa oculta
model_q.add(Dense(64, activation='relu', input_shape=(11,)))

# Segunda capa oculta
model_q.add(Dense(64, activation='relu'))

# Capa de salida para regresión (1 nodo, sin activación)
model_q.add(Dense(1))

# Compilar el modelo con 'mse' y métrica 'mae' (Error Absoluto Medio)
model_q.compile(loss='mse', optimizer='rmsprop', metrics=['mae'])

# Entrenar el modelo (usamos batch_size = 32 para un entrenamiento rápido y eficiente)
model_q.fit(X_train_q, y_train_q, epochs=20, batch_size=32, verbose=1)
```

### 7.3 Evaluación del Modelo de Calidad

Evaluamos en el conjunto de prueba qué tan lejos están, en promedio, nuestras predicciones respecto a la calidad real del vino (usando MAE - Mean Absolute Error).

```python
# Predecir y evaluar en el conjunto de prueba
score_q = model_q.evaluate(X_test_q, y_test_q, verbose=1)
print("MSE (Error Cuadrático Medio) y MAE (Error Absoluto Medio):", score_q)

# Ejemplo de algunas predicciones vs valores reales
pred_ejemplo = model_q.predict(X_test_q[:5])
print("Predicciones:\n", pred_ejemplo.flatten())
print("Valores reales:\n", y_test_q[:5].values)
```
